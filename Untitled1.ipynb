```python
import pandas as pd

df = pd.read_csv("/content/earthquake_1995-2023.csv")

print("rows and coloumns:", df.shape)
df.head()     # show the first 5 rows
```

### Data Cleaning: Initial Inspection

```python
print('--- DataFrame Info ---')
df.info()
```

```python
print('\n--- Percentage of Missing Values ---')
missing_values = df.isnull().sum()
missing_percentage = (missing_values / len(df)) * 100
display(pd.DataFrame({'Missing Count': missing_values, 'Missing Percentage': missing_percentage}).sort_values(by='Missing Percentage', ascending=False))
```

```python
print('\n--- Descriptive Statistics for Numerical Columns ---')
display(df.describe())
```

```python
print('\n--- Descriptive Statistics for Categorical Columns ---')
display(df.describe(include='object'))
```

### Convert `date_time` to datetime objects

```python
# Convert 'date_time' column to datetime objects
df['date_time'] = pd.to_datetime(df['date_time'], format='%d-%m-%Y %H:%M', errors='coerce')

print('\n--- DataFrame Info after date_time conversion ---')
df.info()
```

### Dropping Columns with High Missing Values

```python
columns_to_drop = ['continent', 'alert', 'country']
df.drop(columns=columns_to_drop, inplace=True)
print(f"Columns '{', '.join(columns_to_drop)}' have been dropped.")

print('\n--- Percentage of Missing Values After Dropping Columns ---')
missing_values = df.isnull().sum()
missing_percentage = (missing_values / len(df)) * 100
display(pd.DataFrame({'Missing Count': missing_values, 'Missing Percentage': missing_percentage}).sort_values(by='Missing Percentage', ascending=False))
```

### Final DataFrame Info after Cleaning

```python
print('--- Final DataFrame Info ---')
df.info()
```

### Handle Duplicate Rows

```python
print(f"Number of duplicate rows before dropping: {df.duplicated().sum()}")
df.drop_duplicates(inplace=True)
print(f"Number of duplicate rows after dropping: {df.duplicated().sum()}")
```

### Impute Missing Values for `location`

```python
# Impute missing 'location' values with the mode
location_mode = df['location'].mode()[0]
df['location'].fillna(location_mode, inplace=True)
print(f"Missing values in 'location' after imputation: {df['location'].isnull().sum()}")
```

### Review Columns with High Missing Values

```python
print('\n--- Percentage of Missing Values After Initial Cleaning ---')
missing_values = df.isnull().sum()
missing_percentage = (missing_values / len(df)) * 100
display(pd.DataFrame({'Missing Count': missing_values, 'Missing Percentage': missing_percentage}).sort_values(by='Missing Percentage', ascending=False))
```
