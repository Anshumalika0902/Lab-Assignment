```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib as mpl

df = pd.read_csv("earthquake_1995-2023.csv")

# Count earthquakes for each country
data = df["country"].value_counts().head(10)

# Normalize values for colormap
norm = mpl.colors.Normalize(vmin=data.min(), vmax=data.max())

# Choose colormap
cmap = plt.cm.viridis

# Create colors based on bar height
colors = cmap(norm(data.values))

# Create bar plot
data.plot(kind="bar", color=colors)

# Add colorbar
sm = mpl.cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])

plt.colorbar(sm, ax=plt.gca(), label="Number of Earthquakes")

plt.xlabel("Country")
plt.ylabel("Number of Earthquakes")
plt.title("Top 10 Countries by Number of Earthquakes")
plt.show()
```

The bar plot shows the number of earthquakes recorded in each year from 1995 to 2023. The number of recorded earthquakes varies from year to year. Some years have fewer earthquake records, while other years have more. This visualization helps us easily compare earthquake activity across different years and identify years with relatively higher or lower recorded activity.

```python
import matplotlib.pyplot as plt
from matplotlib.colors import Normalize
import matplotlib.cm as cm

plt.figure(figsize=(10, 6))

data = df['magnitude'].dropna()

counts, bins, patches = plt.hist(
    data,
    bins=30,
    edgecolor='black'
)

norm = Normalize(vmin=counts.min(), vmax=counts.max())

for count, patch in zip(counts, patches):
    patch.set_facecolor(cm.viridis(norm(count)))

sm = cm.ScalarMappable(norm=norm, cmap=cm.viridis)
sm.set_array([])

cbar = plt.colorbar(sm, ax=plt.gca())
cbar.set_label('Frequency')

plt.xlabel('Earthquake Magnitude')
plt.ylabel('Number of Earthquakes')
plt.title('Histogram of Earthquake Magnitudes')
plt.grid(axis='y', alpha=0.3)

plt.show()
```

The histogram shows how earthquake magnitudes are distributed in the dataset. It helps us understand which magnitude ranges occur most frequently and which ranges occur less frequently. The height of each bar represents the number of earthquakes within that magnitude range

```python
import matplotlib.pyplot as plt

country_count = df['country'].value_counts().head(5)

plt.figure(figsize=(7,7))

plt.pie(
    country_count,
    labels=country_count.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Top 5 Countries with Earthquakes')

plt.show()
```

The pie chart shows the percentage distribution of earthquakes according to their magnitude ranges. It helps us easily understand which magnitude range contains the largest proportion of earthquake records and which ranges contain fewer records.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Read CSV file
df = pd.read_csv("/content/earthquake_1995-2023.csv")
# Convert date_time column to date format
df["date_time"] = pd.to_datetime(df["date_time"])

# Extract year
df["year"] = df["date_time"].dt.year

# Count earthquakes for each year
earthquakes_per_year = df["year"].value_counts().sort_index()

# Create line plot
plt.figure(figsize=(10, 5))

plt.plot(
    earthquakes_per_year.index,
    earthquakes_per_year.values,
    marker="o"
)

plt.xlabel("Year")
plt.ylabel("Number of Earthquakes")
plt.title("Number of Earthquakes per Year")
plt.grid(True)

plt.show()
```

The line plot shows the variation in the number of recorded earthquakes from 1995 to 2023. The number of earthquake records changes from year to year, with some years having higher values and others having lower values.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Read CSV file
df = pd.read_csv("/content/earthquake_1995-2023.csv")

# Scatter plot
plt.figure(figsize=(10, 6))

plt.scatter(df["depth"], df["magnitude"])

plt.xlabel("Depth")
plt.ylabel("Magnitude")
plt.title("Earthquake Magnitude vs Depth")
plt.grid(True)

plt.show()
```

The scatter plot shows the relationship between earthquake depth and magnitude. Each point represents one earthquake record. The graph helps us observe whether earthquake magnitude changes consistently with depth.
